```python
import hashlib
import os

def calculate_file_hash(filepath, hash_algorithm='sha256'):
    """Calculates the cryptographic hash of a given file."""
    hasher = hashlib.new(hash_algorithm)
    try:
        with open(filepath, 'rb') as f:

            for chunk in iter(lambda: f.read(4096), b''):
                hasher.update(chunk)
        return hasher.hexdigest()
    except FileNotFoundError:
        return None

filename = "important_document.txt"
initial_content = "This is a very important document that must remain unchanged.\nIt contains sensitive information."

with open(filename, 'w') as f:
    f.write(initial_content)
print(f"Created initial file: '{filename}'")
print("Initial content:\n" + initial_content)

initial_hash = calculate_file_hash(filename)
print(f"\nInitial SHA256 Hash: {initial_hash}")

print("\n--- Simulating file tampering... ---")

with open(filename, 'a') as f:
    f.write("\nSomeone added this line without authorization!")
print(f"File '{filename}' has been modified.")


tampered_hash = calculate_file_hash(filename)
print(f"Tampered SHA256 Hash: {tampered_hash}")


print("\n--- Detecting tampering... ---")
if initial_hash == tampered_hash:
    print("File integrity check passed: The file has NOT been tampered with.")
else:
    print("File integrity check FAILED: The file HAS been tampered with!")


os.remove(filename)
print(f"\nCleaned up file: '{filename}'")
```
