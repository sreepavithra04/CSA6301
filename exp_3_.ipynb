```python
import socket

def scan_port(host, port, timeout=1):
    """Attempts to connect to a single TCP port on a host."""
    try:
        # Create a socket object
        s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        s.settimeout(timeout) # Set a timeout for the connection attempt

        # Attempt to connect to the host and port
        result = s.connect_ex((host, port))
        s.close() # Close the socket

        if result == 0: # Connection successful (port is open)
            return True
        else:
            return False
    except socket.gaierror:
        print(f"Hostname '{host}' could not be resolved. Exiting.")
        return None
    except socket.error:
        print(f"Could not connect to server '{host}'. Exiting.")
        return None

def scan_port_range(target_host, start_port, end_port, timeout=1):
    """Scans a range of TCP ports on the target host and reports open ports."""
    print(f"Scanning {target_host} from port {start_port} to {end_port}...")
    open_ports = []
    for port in range(start_port, end_port + 1):
        status = scan_port(target_host, port, timeout)
        if status is True:
            print(f"Port {port}: Open")
            open_ports.append(port)
        # elif status is False:
        #     print(f"Port {port}: Closed") # Uncomment to see closed ports too
        elif status is None: # An error occurred (e.g., host not resolved)
            break

    if open_ports:
        print(f"\nScan completed. Open ports on {target_host}: {open_ports}")
    else:
        print(f"\nScan completed. No open ports found on {target_host} in the range {start_port}-{end_port}.")

# --- Demonstration ---
# Set your target host and port range
# Example: For testing, you can use a well-known service like google.com (port 80, 443)
# Or, if you have a local server running, you can use '127.0.0.1' or 'localhost'

target_host = "google.com" # CHANGE THIS to your target host
start_port = 79 # Starting port (e.g., 1 for a full scan, but use a smaller range for quick tests)
end_port = 81   # Ending port (e.g., 1024 for common ports, but keep it small for quick tests)

scan_port_range(target_host, start_port, end_port, timeout=0.5)
```
