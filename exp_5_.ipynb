```python
import socket

def resolve_domain_to_ip(domain):
    """
    Resolves a given domain name to its IPv4 address.

    Args:
        domain (str): The domain name to resolve.

    Returns:
        str: The IPv4 address of the domain, or 'N/A' if resolution fails.
    """
    try:
        ip_address = socket.gethostbyname(domain)
        return ip_address
    except socket.gaierror:
        return "N/A - Hostname could not be resolved"
    except Exception as e:
        return f"N/A - Error: {e}"


print("--- Domain Name to IP Resolution Demonstration ---\n")

domain_names = [
    "google.com",
    "github.com",
    "python.org",
    "nonexistentdomain12345.com",
    "facebook.com",
    "wikipedia.org"
]

print("Resolving domain names to IP addresses:\n")
for domain in domain_names:
    ip = resolve_domain_to_ip(domain)
    print(f"Domain: {domain:<30} IP Address: {ip}")
```
